```python
#Using python list
import numpy as np
import time

size = 1000000

l1 = list(range(size))
l2 = list(range(size))

start = time.time()
add = [x+y for x,y in zip(l1,l2)]
end = time.time()
print(add[0:10])

print(f"The total time taken using list is {end - start}")
```

```python
#Using numpy array
import numpy as np
import time

size = 1000000

l1 = np.array((range(size)))
l2 = np.array((range(size)))

start = time.time()
add = l1+l2
end = time.time()
print(add[0:10])

print(f"The total time taken using numpy is {end - start}")
```

```python
#Using python list
import numpy as np
import time

size = 1000000

l1 = list(range(size))

start = time.time()
list_squares = [x ** 2 for x in l1]
end = time.time()
print(list_squares[0:10])

print(f"The total time taken using list is {end - start}")
```

```python
#Using numpy array
import numpy as np
import time

size = 1000000

l1 = np.array((range(size)))

start = time.time()
numpy_squares = l1 ** 2
end = time.time()
print(numpy_squares[0:10])

print(f"The total time taken using numpy is {end - start}")
```
